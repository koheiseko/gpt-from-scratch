```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.pre_tokenizers import ByteLevel
from tokenizers.decoders import ByteLevel as ByteLevelDecoder

import torch

tokenizer = Tokenizer(BPE())
tokenizer.pre_tokenizer = ByteLevel(add_prefix_space=False)
tokenizer.decoder = ByteLevelDecoder()

tokenizer_path = "models/tokenizer.json"
tokenizer = Tokenizer.from_file(tokenizer_path)

print("Tokenizer carregado do disco.")
```

```python
weights = torch.load("models/checkpoints/best_model.pt", weights_only=False)
args = weights["config"]

print("Pesos carregados.")
```

```python
from model import Transformer

device = "cuda"

model = Transformer(args).to(device)
model = torch.compile(model)
model.load_state_dict(weights["model_state_dict"])
```

```python
prompt = "Era uma vez"

context = torch.tensor(
    tokenizer.encode(prompt).ids, dtype=torch.long, device=device
).view(1, -1)

result = model.generate(context, 500, temperature=0.8, do_sample=True, top_k=50)

print(tokenizer.decode(result.view(-1).tolist()))
```
